```python
import fitz
import cv2
import numpy as np
from PIL import Image
import torch

from doctr.io import DocumentFile
from doctr.models import ocr_predictor

from transformers import (
    TrOCRProcessor,
    VisionEncoderDecoderModel
)

# ============================================================
# CONFIG
# ============================================================

PDF_PATH = r"C:\Users\pardh\Downloads\PDP\24-25 Assignment 1\Please upload your assignment file (in .pdf format) (File responses)\22bcs057-3 - KRISHNA JOSHI IIIT Dharwad.pdf"
OUTPUT_FILE = "handwritten_output.txt"

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# ============================================================
# LOAD MODELS
# ============================================================

print("Loading docTR...")

doctr_model = ocr_predictor(
    pretrained=True,
    detect_orientation=True
)

print("Loading TrOCR...")

trocr_processor = TrOCRProcessor.from_pretrained(
    "microsoft/trocr-large-handwritten"
)

trocr_model = VisionEncoderDecoderModel.from_pretrained(
    "microsoft/trocr-large-handwritten"
).to(DEVICE)

print("Models loaded successfully!")

# ============================================================
# IMAGE PREPROCESSING
# ============================================================

def preprocess_for_handwriting(img):
    """
    Improve handwritten OCR quality
    """

    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Denoise
    denoise = cv2.fastNlMeansDenoising(gray)

    # Adaptive threshold
    thresh = cv2.adaptiveThreshold(
        denoise,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        31,
        15
    )

    # Slight dilation
    kernel = np.ones((1, 1), np.uint8)
    dilated = cv2.dilate(thresh, kernel, iterations=1)

    return dilated


# ============================================================
# EXTRACT LINE BOXES USING docTR
# ============================================================

def get_line_boxes(page):
    """
    Extract line bounding boxes from docTR output
    """

    line_boxes = []

    for block in page.blocks:
        for line in block.lines:

            words = line.words

            if len(words) == 0:
                continue

            xs = []
            ys = []

            for word in words:
                geom = word.geometry

                (x1, y1), (x2, y2) = geom

                xs.extend([x1, x2])
                ys.extend([y1, y2])

            x1 = min(xs)
            y1 = min(ys)
            x2 = max(xs)
            y2 = max(ys)

            line_boxes.append((x1, y1, x2, y2))

    return line_boxes


# ============================================================
# TrOCR LINE RECOGNITION
# ============================================================

def trocr_recognize(crop):
    """
    OCR single handwritten line using TrOCR
    """

    pil_img = Image.fromarray(crop).convert("RGB")

    pixel_values = trocr_processor(
        images=pil_img,
        return_tensors="pt"
    ).pixel_values.to(DEVICE)

    generated_ids = trocr_model.generate(
        pixel_values,
        max_new_tokens=128
    )

    text = trocr_processor.batch_decode(
        generated_ids,
        skip_special_tokens=True
    )[0]

    return text


# ============================================================
# MAIN PIPELINE
# ============================================================

print("Reading PDF...")

doc = DocumentFile.from_pdf(PDF_PATH)

print("Running docTR layout analysis...")

result = doctr_model(doc)

final_output = []

# ============================================================
# PROCESS EACH PAGE
# ============================================================

for page_idx, page in enumerate(result.pages):

    print(f"\nProcessing PAGE {page_idx + 1}")

    final_output.append(f"\n===== PAGE {page_idx + 1} =====\n")

    # --------------------------------------------------------
    # Load original page image
    # --------------------------------------------------------

    pdf = fitz.open(PDF_PATH)

    pdf_page = pdf[page_idx]

    pix = pdf_page.get_pixmap(matrix=fitz.Matrix(3, 3))

    img = np.frombuffer(pix.samples, dtype=np.uint8)

    img = img.reshape(pix.height, pix.width, pix.n)

    if pix.n == 4:
        img = cv2.cvtColor(img, cv2.COLOR_RGBA2BGR)
    else:
        img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)

    # --------------------------------------------------------
    # Preprocess image
    # --------------------------------------------------------

    processed = preprocess_for_handwriting(img)

    # --------------------------------------------------------
    # Get line boxes
    # --------------------------------------------------------

    line_boxes = get_line_boxes(page)

    h, w = processed.shape

    page_text = []

    # --------------------------------------------------------
    # OCR EACH LINE
    # --------------------------------------------------------

    for box in line_boxes:

        x1, y1, x2, y2 = box

        # Convert normalized coords to pixels
        x1 = int(x1 * w)
        y1 = int(y1 * h)
        x2 = int(x2 * w)
        y2 = int(y2 * h)

        # Padding
        pad = 8

        x1 = max(0, x1 - pad)
        y1 = max(0, y1 - pad)

        x2 = min(w, x2 + pad)
        y2 = min(h, y2 + pad)

        crop = processed[y1:y2, x1:x2]

        if crop.size == 0:
            continue

        # Skip tiny regions
        if crop.shape[0] < 10 or crop.shape[1] < 30:
            continue

        try:
            text = trocr_recognize(crop)

            text = text.strip()

            if len(text) > 0:
                page_text.append(text)

        except Exception as e:
            print("Line OCR error:", e)

    # --------------------------------------------------------
    # Store page text
    # --------------------------------------------------------

    page_result = "\n".join(page_text)

    print(page_result)

    final_output.append(page_result)

# ============================================================
# SAVE OUTPUT
# ============================================================

with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    f.write("\n".join(final_output))

print(f"\nSaved to {OUTPUT_FILE}")
```
